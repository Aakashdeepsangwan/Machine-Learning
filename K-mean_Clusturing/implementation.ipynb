```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
import pandas as pd
import numpy as np
%matplotlib inline
```

## Make_blobs is fake 2D point data made so we can practice clusturing
- 3 centres mean in the starting 3 points are initialised and 1000 points are dropped inside the graph
- Y is like the answer key, it stores in which Centriod a particular point belong in X 


```python
X,y = make_blobs(n_samples=1000, centers = 3, n_features=2)
```

```python
plt.scatter(X[:,0],X[:,1], c=y)
```

```python
## Standardization
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.33, random_state = 42)
X_train_scaled = scaler.fit_transform(X_train)
x_test_scaled = scaler.transform(X_test) # To prevent Data Leakage
```

```python
X_train_scaled
```

```python
print(X[0])
print(X_train_scaled[0])
```

```python
x_test_scaled.shape
```

```python
x_test_scaled
```

```python
from sklearn.cluster import KMeans
# Applying Elbow Method
wcss = [] # Distance square plus all the Euclidian distances related to a particular centriod
for k in range(1, 11) : 
    kmeans = KMeans(n_clusters=k, init = "k-means++")
    kmeans.fit(x_test_scaled)
    wcss.append(kmeans.inertia_)
```

```python
wcss
```

```python
# In order to determine the Exact value of K - Elbow Curve
plt.plot(range(1,11),wcss)
plt.xticks(range(1,11)) # ??
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.show()
```

## Using K elbow method, we clearly knows we only need 2 centriods to Cluster this data

```python
kmeans = KMeans(n_clusters=3, init="k-means++")
kmeans.fit_predict(X_train_scaled)
```

```python
y_pred = kmeans.predict(x_test_scaled)
y_pred
```
